```python
# Response Context: OpenAI Responses API with conversation threading.
#
# Demonstrates:
#   - client.responses.create()      -- the Responses API (not Chat Completions)
#   - previous_response_id           -- server-side conversation memory
#   - client.responses.retrieve()    -- recall a stored response by ID
#
# Designed for Google Colab. If running locally, replace the userdata
# call with:  os.environ["OPENAI_API_KEY"] = "your-key-here"

from openai import OpenAI
from google.colab import userdata
import os
```

```python
OPEN_API_KEY = userdata.get('OPENAI_API_KEY')
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
```

```python
client = OpenAI()
```

```python
# ---- First Response ----
# Uses the Responses API (not Chat Completions).
response = client.responses.create(
    model="gpt-4o-mini",
    input="tell me a joke",
)
print(response.output_text)
```

```python
# ---- Retrieve a stored response by ID ----
response_id = response.id
response_recall = client.responses.retrieve(response_id)
print(response_recall.output_text)
print(response_recall)
```

```python
# ---- Second Response (chained via previous_response_id) ----
# The server recalls the full conversation; no need to resend history.
second_response = client.responses.create(
    model="gpt-4o-mini",
    previous_response_id=response.id,
    input=[{"role": "user", "content": "explain why this is funny."}],
)
print(second_response.output_text)
```

```python
# ---- Retrieve second response ----
response_id = second_response.id
response_recall = client.responses.retrieve(response_id)
print(response_recall.output_text)
print(response_recall)
```

```python
# ---- Third Response (chained again) ----
# The full thread is preserved server-side: joke -> explanation -> summary.
third_response = client.responses.create(
    model="gpt-4o-mini",
    previous_response_id=second_response.id,
    input=[{"role": "user", "content": "Summarise the actions of this thread."}],
)
print(third_response.output_text)
```

```python
# ---- Retrieve third response ----
response_id = third_response.id
response_recall = client.responses.retrieve(response_id)
print(response_recall.output_text)
print(response_recall)
```
